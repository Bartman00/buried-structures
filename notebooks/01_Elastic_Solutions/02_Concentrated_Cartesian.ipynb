## Cylindrical coordinates

The point load Boussinesq solution is given in cylindrical rather than cartesian coordinates.

![Boussinesq formulation](./images/point_normal.png)

This notebook is to explore the converting them.

![Cylindrical coordinates](./images/cylindrical_coordinates.png)

```python
def find_extreme(iterable, func=max) -> float|None:
    """
    Recursively find the min or max in nested iterables.
    :param iterable: The nested collection (list, tuple, etc.)
    :param func: The built-in min or max function
    """
    flat_elements = []
    
    for item in iterable:
        # Check if the item is a nested iterable (excluding strings if desired)
        if isinstance(item, (list, tuple, set)):
            # Recursive call to drill into the nested structure
            inner_extreme = find_extreme(item, func)
            if inner_extreme is not None:
                flat_elements.append(inner_extreme)
        else:
            # Base case: item is a single value (int, float, etc.)
            flat_elements.append(item)
            
    return func(flat_elements) if flat_elements else None

# Example usage:
nested_data = [1, [5, [10, -2]], 8, [0]]
print(f"Maximum: {find_extreme(nested_data, max)}") # Output: 10
print(f"Minimum: {find_extreme(nested_data, min)}") # Output: -2
```

```python
def radial(p: Point_3d, v: float = 0.30) -> float:
    # Return radial stress component from Boussinesq

    o = Origin()
    r = o.dr(p) # Distance on xy plain

    if r < 1e-6:
        # Which direction would this even go?
        return 0

    R = o.distance(p) # Total distance

    return (-1/(2*math.pi*R**2)) * (-3*r**2*p.z()/(R**3) + (1-2*v)*R/(R+p.z()))

print(radial(Point_3d(0, 0, 10), 0.30))

# Tangental
def tangential(p: Point_3d, v: float = 0.30) -> float:
    # Return radial stress component from Boussinesq

    o = Origin()
    r = o.dr(p) # Distance on xy playne
    R = o.distance(p) # Total distance

    if r < 1e-6:
        # Which direction would this even go?
        return 0


    # print(f"Point: \n{p}")
    # print(f"z: {p.z()}")
    # print(f"R: {R}")
    # print(f"term: {p.z()/R + R/(R+p.z())}")

    return -(1-2*v)/(2*math.pi*R**2) * (p.z()/R - R/(R+p.z()))


print(tangential(Point_3d(0, 0, 10), 0.30))
```

```python
# Volume plot of radial stress
X, Y, Z = np.mgrid[12:48:40j, 12:48:40j, 18:48:40j]
x, y, z = X.flatten(), Y.flatten(), Z.flatten()
v = 0.3

values = [radial(Point_3d(x=ix, y=iy, z=iz), v) for ix, iy, iz in zip(x, y, z)]

```

```python
# Plot

def volume_fig(x, y, z, values, title, colorbar_title) -> go.Figure:
    # Generic function for generating volume plots
    isomin, isomax = find_extreme(values, min), find_extreme(values, max)
    fig = go.Figure(go.Volume(
        x=x, 
        y=y, 
        z=z,
        value=values,
        isomin=isomin,
        isomax=isomax,
        colorscale='Turbo',
        showscale=True,
        surface_count=20,
        colorbar=dict(title=colorbar_title),
        opacity=0.3,
    ))

    fig.update_layout(
        title=title,
        width=600,
        height=600,
        scene=dict(
            xaxis_title='X',
            yaxis_title='Y',
            zaxis_title='Z',
            zaxis=dict(autorange='reversed')  # reverse z so depth is shown accordingly
        )
    )

    return fig

fig = volume_fig(x, y, z, values, "Radial_Stress", "Stress")
fig.show()
```

```python

values_tan = [tangential(Point_3d(x=ix, y=iy, z=iz), v) for ix, iy, iz in zip(x, y, z)]

# Plot


fig_tan = volume_fig(x, y, z, values_tan, "Tangential Stress", "Stress")
fig_tan.show()
```

## Applying the trigonometry

Looking again at the plan view:

![Cylindrical coordinates](./images/cylindrical_coordinates.png)

We can see that when theta = 0, sigma_r = sigma_x, and sigma_theta = sigma_y.

However, we can't apply the expected transform of:

\begin{align*}
\sigma_x &= \sigma_r \cos\theta + \sigma_{\theta} \sin\theta\\
\sigma_y &= \sigma_r \sin\theta + \sigma_{\theta} \cos\theta
\end{align*}

This is because all these stresses aren't really vectors. The are scalers and at theta = 180 degrees, sigma_theta = sigma_x just like at theta = 0 degrees.

So instead we use the absolute value of sin and cos:

\begin{align*}
\sigma_x &= \sigma_r abs(\cos\theta) + \sigma_{\theta} abs(\sin\theta)\\
\sigma_y &= \sigma_r abs(\sin\theta) + \sigma_{\theta} abs(\cos\theta)
\end{align*}

```python
x_degrees = np.linspace(0, 360, 360)
x_rad = [math.radians(x) for x in x_degrees]

y_sin = [math.sin(x) for x in x_rad]
y_cos = [math.cos(x) for x in x_rad]
y_a_sin = [abs(x) for x in y_sin]
y_a_cos = [abs(x) for x in y_cos]

fig = make_subplots(rows=2, cols=1, subplot_titles=('Sin / Cos', 'Absolute Sin / Cos'))

fig.add_trace(go.Scatter(x=x_degrees, y=y_sin, name='Sin'), row=1, col=1,)
fig.add_trace(go.Scatter(x=x_degrees, y=y_cos, name='Cos'), row=1, col=1,)

fig.add_trace(go.Scatter(x=x_degrees, y=y_a_sin, name='abs(Sin)'), row=2, col=1,)
fig.add_trace(go.Scatter(x=x_degrees, y=y_a_cos, name='abs(Cos)'), row=2, col=1,)
              
fig.show()
```

```python
def cylidrical_chunck(r1, r2, t1, t2, npoints: int =20):
    # Returns a circular slice
    assert (npoints % 2 == 0), "npoints needs to be even"
    assert (npoints >= 4), "npoints needs to be >= 4"

    n = int(npoints/2)

    x, y = [], []

    r = [r1] * n
    r.extend([r2] * n)
    print(f"{r=}")

    t = np.linspace(t1, t2, n).astype(float).tolist()
    print(f"{t=}")

    x = [ir*math.cos(math.radians(it)) for ir, it in zip(r, t)]
    y = [ir*math.sin(math.radians(it)) for ir, it in zip(r, t)]

    return x, y

points_x, points_y = cylidrical_chunck(1, 2, 30, 60)
print(f"{points_x}")
print(f"{points_y}")
```

```python
def sigma_x(p: Point_3d, v: float = 0.30) -> float:
    # X direction component of cylindrical stress
    theta = math.atan2(p.y(), p.x())

    r, t = radial(p, v), tangential(p, v)

    return r*abs(math.cos(theta)) + t*abs(math.sin(theta))

def sigma_y(p: Point_3d, v: float = 0.30) -> float:
    # Y direction component of cylindrical stress
    theta = math.atan2(p.y(), p.x())

    r, t = radial(p, v), tangential(p, v)

    return r*abs(math.sin(theta)) + t*abs(math.cos(theta))
```

```python

values_x = [sigma_x(Point_3d(x=ix, y=iy, z=iz), v) for ix, iy, iz in zip(x, y, z)]
print(f"Min x stress: {min(values_x)}")
print(f"Max x stress: {max(values_x)}")

# Plot
fig_x = volume_fig(x, y, z, values_x, "x Stress", "Stress")
fig_x.show()
```

```python

values_y = [sigma_y(Point_3d(x=ix, y=iy, z=iz), v) for ix, iy, iz in zip(x, y, z)]
print(f"Min y stress: {min(values_y)}")
print(f"Max y stress: {max(values_y)}")


# Plot
fig_y = volume_fig(x, y, z, values_y, "y Stress", "Stress")
fig_y.show()
```

## Important locations

Of particular interest are the effects at y = 48" and y = 60"

This is because this is the distance from a wheel load in an adjacent lane and 2nd wheel line used in a 2D frame analysis. Some additional curves are drawn for sigma_x below at these locations and at x intervals from 0' (Directly over) to 10' (in longitudinal direction)

```python
# y = 48"
y_single = 48

x_48 = np.linspace(0, 120, 10)
z_48 = np.linspace(18,120, 10)
# print(x_48)

fig = go.Figure()
fig = make_subplots(rows=2, cols=1, subplot_titles=('sigma_x', 'radial'))

# fig.add_trace(go.Scatter(x=x_degrees, y=y_sin, name='Sin'), row=1, col=1,)
# fig.add_trace(go.Scatter(x=x_degrees, y=y_cos, name='Cos'), row=1, col=1,)

theta = [math.atan2(y_single, ix) for ix in x_48]

for iz in z_48:
    y_48_values = [sigma_x(Point_3d(ix, y_single, iz), v) for ix in x_48]
    radial_48_values = [radial(Point_3d(ix, y_single, iz), v)*math.cos(it) for ix, it in zip(x_48, theta)]
    # print(y_48_values)

    fig.add_trace(
        go.Scatter(
            x=x_48, 
            y=y_48_values,
            name=f"z={iz:.2f}",
            
        ),row=1,col=1
    )
    

    fig.add_trace(
        go.Scatter(
            x=x_48, 
            y=radial_48_values,
            name=f"z={iz:.2f}",
            
        ),row=2,col=1
    )
    
    
        

fig.update_layout(
    title='Sigma_x at y=48',
    xaxis_title='x (in)',
    yaxis_title='Sigma_x (ksi)',
    template='plotly_white',
    width=800,
    height=800
    
)

fig.show()
```

```python
fig_solid = go.Figure()


print(x_48)
print(z_48)

color='red'

theta = [math.atan2(y_single, ix) for ix in x_48]

value = []
radial_values = []
for ix ,it in zip(x_48, theta):
    new_row = []

    new_radial_row = []

    for iz in z_48:
        point = Point_3d(x=ix, y=y_single, z=iz)
        new_row.append(sigma_x(point, v))
        new_radial_row.append(radial(point, v)*math.cos(it))
    value.append(new_row)
    radial_values.append(new_radial_row)

fig_solid.add_trace(go.Surface(
    z=value,
    x=x_48, y=z_48,
    surfacecolor=np.zeros((len(x_48), len(z_48))),
    colorscale=[[0, color], [1, color]],
    opacity=0.5,
    name=f"sigma_x",
    showscale=False
))

radial_color = 'green'
fig_solid.add_trace(go.Surface(
    z=radial_values,
    x=x_48, y=z_48,
    surfacecolor=np.zeros((len(x_48), len(z_48))),
    colorscale=[[0, radial_color], [1, radial_color]],
    opacity=0.5,
    name=f"radial component",
    showscale=False
))

# radial_values = [radial(Point_3d(ix, y_single, iz), v)*math.cos(it) for ix, it in zip(x_48, theta)]
    # print(y_48_values)


fig_solid.update_layout(
    showlegend=True,
    title="x-direction with and without tangential at y=48. Green is radial only component",
    # 
    width=800,
    height=600,
    scene=dict(
        xaxis_title="x (in)",
        yaxis_title="z (in)",
        zaxis_title="Pressure (ksi)"
    )
)

fig_solid.show()
```

## Conclusions

Converting from the cylindrical coordinates in Poulos & Davis takes an extra step because the equations are outputting stresses rather than actual vectors.

This can be accomplished by taking the absolute value of the radial and tangential stress components.

For stresses at locations z > 18" you'll typically see positive compressive forces in both the x & y direction with some small negative stresses (tension) directly under the laods.

At y = 48", and 60" which are used when designing or rating with AASHTO, at higher depths you can see some small negative stresses at deep depths.

Compared to ignoring the tangetial component, there will be a significant increase in x & y stresses near the surface with smaller decreases and even net tension at further horizontal distances and depths.
