```python
# Define loads
wheel_magnitude = 16
poisson = 0.30
length, width = 10, 20
load_point = Point_3d(0, 0, 0)
area_magnitude = wheel_magnitude / (length * width)

rectangular_load = Rectangular_Load(load_point, area_magnitude, length, width)

l_divisions = 10
w_divisions = 20

approx_load = Rectangular_Approximation(rectangular_load, l_divisions, w_divisions, poisson=0.3)
```

```python
# Plot X, Y, Z stresses in a volume
X, Y, Z = np.mgrid[0:60:40j, 0:60:40j, 6:60:40j]
x, y, z = X.flatten(), Y.flatten(), Z.flatten()

x_stress = [approx_load.stress_x(Point_3d(x=ix, y=iy, z=iz)) for ix, iy, iz in zip(x, y, z)]
y_stress = [approx_load.stress_y(Point_3d(x=ix, y=iy, z=iz)) for ix, iy, iz in zip(x, y, z)]
z_stress = [approx_load.stress_z(Point_3d(x=ix, y=iy, z=iz)) for ix, iy, iz in zip(x, y, z)]
```

```python
# Plot X
fig_x = volume_plot(x, y, z, x_stress, title="X-direction Stress")
fig_x.show()
```

```python
# Plot Y
fig_y = volume_plot(x, y, z, y_stress, title="Y-direction Stress")
fig_y.show()
```

```python
# Plot Z
fig_z = volume_plot(x, y, z, z_stress, title="Z-direction Stress")
fig_z.show()
```
