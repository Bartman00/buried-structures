### Purpose

The purpose of this notebook is to generate imshow plots in a manner I'll be using throughout this library.

### Included:

1. Basic imshow plot with some formatting explored. These are useful for taking a slice of fill and seeing how pressures vary over the area of the slice.
2. How to generate a list of slices. I use this for keeping the same x & y coordinates, but taking different slices varying the z coordinate.
3. How to animate. Similar to generating a list of slices, this allows varying one coordinate to see how it changes.

```python
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np
from math import pi
from buried_structures import Point_3d, Origin
import xarray as xr
print("Import tests finished")
```

```python
def bous(p: Point_3d) -> float:
    # Simplified Boussinesq problem with hard coded
    # inputs that only returns downward pressure.
    # Used for demonstrating plotly charts work as
    # intended.

    o = Origin()
    R = p.distance(o)
    R = R if R > 0 else 0.001 # Avoid singularities
    z = p.dz(o)
    P = 1.0

    # print(f"{R=}, {z=}")
    return 3*P*(z**3) / (2 * pi * (R**5))

p = Point_3d(1,1,1)
print(f"pressure = {bous(p)}")
```

```python
# Generate imshow plot of 

# Setup linspaces. Set more to make good contrast images
from_x, to_x = 0, 5
from_z, to_z = 10, 11
x_arr = np.linspace(from_x, to_x, 101)
z_arr = np.linspace(from_z, to_z, 101)

pressure = []
for z in z_arr:
    new_row = []
    for x in x_arr:
        # print(f"{x=}, {z=}")
        t_pressure = bous(Point_3d(x, 0, z))
        new_row.append(t_pressure)

    pressure.append(new_row)

# print("[")
# for row in pressure:
#     print(row)
# print("]")

fig = px.imshow(pressure, color_continuous_scale="Turbo")
fig.show()
```

```python
# Gemini generated example:


# Define coordinates using linspace (start, stop, number of points)
x_coords = np.linspace(0, 10, 5)   # 5 points from 0 to 10 inches
y_coords = np.linspace(0, 6, 4)    # 4 points from 0 to  6 inches

# Generate matching dummy pressure data (5x4 array)
pressure_data = np.random.uniform(14.7, 30.0, size=(5, 4))

# Create the DataArray
pressure_da = xr.DataArray(
    data=pressure_data,
    dims=("x", "y"),
    coords={
        "x": ("x", x_coords, {"units": "inches"}),
        "y": ("y", y_coords, {"units": "inches"}),
    },
    name="pressure",
    attrs={"units": "psi"}
)

print(pressure_da)
fig = px.imshow(pressure_da, color_continuous_scale='Turbo', aspect='equal')
fig.show()
```

```python
# Plot with xarray using our Example:


# Define coordinates using linspace (start, stop, number of points)
print(f"{from_x=}, {to_x}")
print(f"{from_z=}, {to_z}")
x_coords = np.linspace(from_x, to_x, 101)   # 5 points from 0 to 10 inches
z_coords = np.linspace(from_z, to_z, 101)    # 4 points from 0 to  6 inches


# Generate matching dummy pressure data (5x4 array)
# pressure_data = np.random.uniform(14.7, 30.0, size=(5, 4))
pressure_data = []

for z in z_coords:
    new_row = []
    for x in x_coords:
        point = Point_3d(x, 0, z)
        new_row.append(bous(point))
    
    pressure_data.append(new_row)

# Create the DataArray
pressure_da = xr.DataArray(
    data=pressure_data,
    dims=("z (in)", "x (in)"),
    coords={
        "z (in)": ("z (in)", z_coords, {"units": "inches"}),
        "x (in)": ("x (in)", x_coords, {"units": "inches"}),
    },
    name="pressure",
    attrs={"units": "ksi"}
)

# print(pressure_da)

fig = px.imshow(pressure_da, color_continuous_scale='Turbo', aspect='equal')
fig.show()
```

```python
# Add additional Formatting options for title & axes
fig.update_layout(title="Simple Pressure Distribution")
fig.show()
```

```python
# 3D Image Version. 


# Constant x & y while the slices vary the z coordinate.

# Define coordinates using linspace (start, stop, number of points)
x_coords = np.linspace(from_x, to_x, 101)
y_coords = np.linspace(from_x, to_x, 101) # Same as x
z_coords = np.linspace(from_z, to_z, 6)


# Generate matching dummy pressure data (5x4 array)
# pressure_data = np.random.uniform(14.7, 30.0, size=(5, 4))
pressure_data = []

for z in z_coords:
    new_row = []

    for y in y_coords:
        new_column = []
        for x in x_coords:
            point = Point_3d(x, y, z)
            new_column.append(bous(point))
        new_row.append(new_column)
    
    pressure_data.append(new_row)

# Create the DataArray
pressure_da = xr.DataArray(
    data=pressure_data,
    dims=("z (in)", "y (in)", "x (in)"),
    coords={
        "z (in)": ("z (in)", z_coords, {"units": "inches"}),
        "y (in)": ("y (in)", y_coords, {"units": "inches"}),
        "x (in)": ("x (in)", x_coords, {"units": "inches"}),
    },
    name="pressure",
    attrs={"units": "ksi"}
)

min_pressure = pressure_da.min().item()
max_pressure = pressure_da.max().item()

# print(pressure_da)
fig = px.imshow(pressure_da, 
                facet_col="z (in)", facet_col_wrap=3,
                zmin=min_pressure, zmax=max_pressure, 
                color_continuous_scale='Turbo', aspect='equal',
                height=1000, width=1000)
fig.update_layout(title="Simple Pressure Distribution with Slices")
fig.show()
```

```python
# Animate

# Define coordinates using linspace (start, stop, number of points)
x_coords = np.linspace(from_x, to_x, 101)
y_coords = np.linspace(from_x, to_x, 101) # Same as x
z_coords = np.linspace(from_z, to_z, 101)


# Generate matching dummy pressure data (5x4 array)
# pressure_data = np.random.uniform(14.7, 30.0, size=(5, 4))
pressure_data = []

for z in z_coords:
    new_row = []

    for y in y_coords:
        new_column = []
        for x in x_coords:
            point = Point_3d(x, y, z)
            new_column.append(bous(point))
        new_row.append(new_column)
    
    pressure_data.append(new_row)

# Create the DataArray
pressure_da = xr.DataArray(
    data=pressure_data,
    dims=("z (in)", "y (in)", "x (in)"),
    coords={
        "z (in)": ("z (in)", z_coords, {"units": "inches"}),
        "y (in)": ("y (in)", y_coords, {"units": "inches"}),
        "x (in)": ("x (in)", x_coords, {"units": "inches"}),
    },
    name="pressure",
    attrs={"units": "ksi"}
)

min_pressure = pressure_da.min().item()
max_pressure = pressure_da.max().item()

# print(pressure_da)
fig = px.imshow(pressure_da, animation_frame='z (in)', 
                zmin=min_pressure, zmax=max_pressure, 
                color_continuous_scale='Turbo', aspect='equal',
                width=800, height=800)

# Speed controls
fig.layout.updatemenus[0].buttons[0].args[1]["frame"]["duration"] = 20
fig.layout.updatemenus[0].buttons[0].args[1]["transition"]["duration"] = 00
fig.update_layout(title="Animated Simple Pressure Distribution")
fig.show()
```
